```python
from sklearn.ensemble import RandomForestClassifier as RFC
from sklearn.feature_selection import SelectFromModel
```

```python
RFC_ = RFC(n_estimators=10, random_state=42)
```

```python
# 如何确定threshold呢
X_embedded = SelectFromModel(RFC_, threshold=0.005).fit_transform(X, Y)
X_embedded.shape
```

```python
max(RFC_.fit(X, Y).feature_importances_)
```

```python
# 找到最大的importance后绘制不同阈值的学习曲线
thresholds = np.linspace(0, max(RFC_.fit(X, Y).feature_importances_), 15)
```

```python
score = []
for threshold in thresholds:
    x_embedded = SelectFromModel(RFC_, threshold=threshold).fit_transform(X, Y)
    s = cross_val_score(RFC_, x_embedded, Y, cv=5).mean()
    score.append(s)
```

```python
plt.plot(thresholds, score)
plt.show()
```

```python
# 根据学习曲线选择一个合适的threshold
# 这里选两个边界值
selected_thresholds = thresholds[1:3]
score2 = []
for t in np.linspace(selected_thresholds[0], selected_thresholds[1], 6):
    x_embedded = SelectFromModel(RFC_, threshold=t).fit_transform(X, Y)
    s = cross_val_score(RFC_, x_embedded, Y, cv=5).mean()
    score2.append(s)
```

```python
plt.plot(np.linspace(selected_thresholds[0], selected_thresholds[1], 6), score2)
plt.show()
```

```python
# 在调整一下RFC的n_estimators参数
max(score2)
```

```python
RFC_1 = RFC(n_estimators=100, random_state=42)
x_embedded = SelectFromModel(RFC_1, threshold=selected_thresholds[0]).fit_transform(X, Y)
cross_val_score(RFC_1, x_embedded, Y, cv=5).mean()
```
